```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

***checking*** **outliers**

```
num_feature = df.select_dtypes(include = "number").columns
```

```
for i in num_feature:
  q1 = np.percentile(df[i],25)
  q3 = np.percentile(df[i],75)
  iqr = q3-q1
  upper = q3 + (1.5*iqr)
  loower = q1 -( 1.5 * iqr)
  meann = df[i] .mode()
  df.loc[(df[i] < loower) | (df[i] > upper), i] = meann
```

```
num_cols = df.select_dtypes(include = "number")
num_cols.skew()
#near to zero so well skew
```

```
buck1 = df["Country"].value_counts().head(10).index.tolist()
buck1
df["Country"].value_counts()
```

```
def group_co(country):
  if country in buck1 :
    return country
  else :
    return "Other"
```

```
df["Country"] = df["Country"].apply(group_co)
```

```
#skew clm like study ours will be log transformation
#ordinal_clm will be get label encoding like stress level
#oter numerical will get standard scaler
```

```
df.dropna(inplace=True)
skewd_col= ["Study_Hours"]
num_cols = [
    "Age",
    "Avg_Daily_Usage_Hours",
    "Daily_Unlocks",
    "Physical_Activity_Hours",
    "Sleep_Hours_Per_Night"
]
ordinal_col = ["Stress_Level"]
nomal_col = [
    "Gender",
    "Country",
    "Academic_Level",
    "Most_Used_Platform",
    "Purpose_Of_Use"
]
x = df[skewd_col + num_cols + ordinal_col + nomal_col]
y = df["Mental_Health_Score"]

from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state = 42)
```

```
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer,StandardScaler,OrdinalEncoder,OneHotEncoder
from sklearn.compose import ColumnTransformer

#skew pipeline
skew_pipeline = Pipeline(steps = [('log_transformer',FunctionTransformer(np.log1p)),
                                  ("scale",StandardScaler()),
     ])


#numeric feature

numeric_feature = Pipeline(steps= [
   ( "scale",StandardScaler())

])


#ordinal features
ordinal_feature = Pipeline(steps=[
  (  "encoder",OrdinalEncoder(categories=[["Low","Medium","High","Very High"]]))

])

#normal_feature
normal_feature = Pipeline(steps = [
   ( "encoder",OneHotEncoder(handle_unknown = "ignore"))
])

preprocessor = ColumnTransformer(transformers = [
    ("skew",skew_pipeline,skewd_col),
    ("numeric",numeric_feature,num_cols),
    ("ordinal",ordinal_feature,ordinal_col),
    ("cat",normal_feature,nomal_col)


])
```

```
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score

lr_pipeline = Pipeline(steps = [
    ("processor" , preprocessor),
    ("regressor", LinearRegression())
])
lr_pipeline.fit(x_train,y_train)
y_pred = lr_pipeline.predict(x_test)
print(r2_score(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
```

```
from sklearn.ensemble import RandomForestRegressor
rf_pipeline = Pipeline(steps=[
    ("processor",preprocessor),
    ("regressor",RandomForestRegressor())
])
rf_pipeline.fit(x_train,y_train)
y_pred = rf_pipeline.predict(x_test)
print(r2_score(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
```

```
param_grid = {
    "regressor__n_estimators": [100, 200, 500],
    "regressor__max_depth": [5, 10, 15],
    "regressor__min_samples_split": [2, 5, 10],
    "regressor__min_samples_leaf": [1, 2, 4]
}
```

```
from sklearn.model_selection import RandomizedSearchCV
model = RandomizedSearchCV(
    estimator = rf_pipeline,
    param_distributions = param_grid,
    n_iter = 15,
    cv = 5,
    scoring = "r2",
    random_state= 42 ,

)
model.fit(x_train,y_train)
y_pred = model.predict(x_test)
print(r2_score(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
```
